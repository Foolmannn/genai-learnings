```python
from abc import ABC , abstractmethod
```

```python
class Runnable(ABC):

    @abstractmethod
    def invoke(input_data):
        pass
```

```python
import random
class FakeLLM(Runnable):

    def __init__(self):
        print('llm created')

    def invoke(self,prompt):
        responseList = [
            "Hello this is random response 1 ",
            "Cricket is another response",
            "AI is cool is another response"
        ]
        return {'response': random.choice(responseList)}
# predict is not removed for legacy 
    def predict(self,prompt):
        responseList = [
            "Hello this is random response 1 ",
            "Cricket is another response",
            "AI is cool is another response"
        ]
        return {'response': random.choice(responseList),"message":"this method is deprecated use invoke "}

    
```

```python
class FakePromptTemplate:
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables=input_variables
    def invoke(self, input_dict):
        return self.template.format(**input_dict)


    def format(self,input_dict):
        return self.template.format(**input_dict)

    
```

# so by using the OOP abstraction and inheritance we have implemented standardaization. 

Adding the str output parser 


```python
class FakeStrOutputParser(Runnable):
    def __init__(self):
        pass
    def invoke(self,input_data):
        return input_data['response']
    
```

# So now we create the new class that combines the two runnables . 

```python
class RunnableConnector(Runnable):

    def __init__(self, runnable_list):
        self.runnable_list = runnable_list
    def invoke(self,input_data):

        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)  # output of previous chain act as input 
        return input_data # returning final step output 
```

```python
llm = FakeLLM()
```

```python
template = FakePromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=['length','topic']
)
```

```python
chain = RunnableConnector([template,llm])
```

```python
chain.invoke({'length':'short',"topic":"Nepal"})
```

```python
parser = FakeStrOutputParser()
```

```python
chain1= RunnableConnector([template,llm,parser])
```

```python
chain1.invoke({'length':'short',"topic":"Nepal"})
```
