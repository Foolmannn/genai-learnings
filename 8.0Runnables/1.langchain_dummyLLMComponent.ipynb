Dummy classes of the langchain to understand runnables 

DUMMY LLM CLASS

```python
import random
class FakeLLM:

    def __init__(self):
        print('llm created')

    def predict(self,prompt):
        responseList = [
            "Hello this is random response 1 ",
            "Cricket is another response",
            "AI is cool is another response"
        ]
        return {'response': random.choice(responseList)}

    
```

```python
llm = FakeLLM()
```

```python
llm.predict("what is the capital of Nepal ?")
```

DUMMY PROMPT CLASS

```python
class FakePromptTemplate:
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables=input_variables

    def format(self,input_dict):
        return self.template.format(**input_dict)

    
```

```python
template = FakePromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=['length','topic']
)

```

```python
prompt  =template.format({'length':'short', 'topic': 'Nepal'})
```

```python
llm = FakeLLM()
```

```python
llm.predict(prompt)
```

# this demonstrate a simple llm and prompt class like in langchain . 

# Now creating a chain like langchain 

```python
class FakeLLMChain:

    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self,input_dict):

        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)
        return result['response']

```

```python
llm = FakeLLM()
template = FakePromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=['length','topic']
)
```

```python
chain = FakeLLMChain(llm,template)
```

```python
chain.run({'length':'short',"topic":"Nepal"})
```
